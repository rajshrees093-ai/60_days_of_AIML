```python
##first step is to import pandas
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame(np.arange(0,20).reshape(5,4),index=['Row1','Row2','Row3','Row4','Row5'],columns=["Column1","Column2","Column3","Column4"])
df.head()
```

```python
df.to_csv('test1.csv', index=False)
```

```python
#accessing the elements
#1. .loc 2. .iloc
#In pandas, .loc and .iloc are used to select specific rows and columns from a DataFrame, but they use entirely different indexing methods.The core difference is that .loc is label-based, while .iloc is integer-position-based.

df.loc['Row1']
```

```python
type(df.loc['Row1'])
```

```python
df.iloc[:,:]
```

```python
#take elements from column 2 only
df.iloc[:,1:2]
```

```python
type(df.iloc[:,1:2])
```

```python
#converting df into arrays
df.iloc[:, 1:].values.shape
```

```python
from io import StringIO, BytesIO
```

```python
data=('col1,col2,col3\n'
      'x,y,1\n'
      'a,b,2\n'
      'c,d,3')
```

```python
type(data)
```

```python
pd.read_csv(StringIO(data))
```

```python
#read from specific columns
df=pd.read_csv(StringIO(data),usecols=lambda x: x.upper() in ['COL1', 'COL3'])
df.head()
```

```python
df.to_csv('Test.csv')
```
