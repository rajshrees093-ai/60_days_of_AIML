# Map function in python

```python
def even_or_odd(num):
  if num%2==0:
    return "even"
  else:
    return "odd"
```

```python
even_or_odd(23)
```

```python
lst=[1,2,3,4,5,6,7,8,9,24,56,78]
```

```python
list(map(even_or_odd, lst))
```

# Filter function in Python

```python
def even(num):
  if num%2==0:
    return True
```

```python
lst=[1,2,3,4,5,6,7,8,9,8]
```

```python
list(filter(even,lst))
```

```python
list(filter(lambda num:num%2==0, lst))
```

```python
list(map(lambda num:num%2==0, lst))
```

# List Comprehension

List comprehension provide a concise way to create lists. It consists of brackets containing an expression followed by a for clause, then zero or more for or if clauses. The expression can be anything, meaning you can put in all kinds of objects in lists.

```python
lst1=[]
def lst_square(lst):
  for i in lst:
    lst1.append(i*i)
  return lst1
```

```python
lst_square([1,2,3,4,5,6,7,8,9])
```

```python
lst=[1,2,3,4,5,6,7,8,9]
```

```python
[i*i for i in lst if i%2==0]#list comprehension
```
