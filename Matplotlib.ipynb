```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
import pandas as pd
```

```python
#simple examples

x=np.arange(0,10)
y=np.arange(11,21)
```

```python
#plotting using matplotlib
#plt scatter

plt.scatter(x,y,c='g')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Graph in 20')

```

```python
y=x*x
```

```python
#plt plot
plt.plot(x,y, 'r*', linewidth=2, linestyle='dashed', markersize=12)
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('2-D diagram')
```

```python
#creating subplots
plt.subplot(2,2,1)
plt.plot(x,y,'r')
plt.subplot(2,2,2)
plt.plot(x,y,'g*--')
plt.subplot(2,2,3)
plt.plot(x,y, 'bo')





```

```python
x=np.arange(0,4*np.pi,0.1)
y=np.sin(x)
plt.title("Sine wave form")

#plot the points using matplotlib
plt.plot(x,y)
plt.show()
```

```python
#subplot()
#ompute the x and y cordinates for points on sine and cosine curves

x=np.arange(0, 5*np.pi, 0.1)
y_sin=np.sin(x)
y_cos=np.cos(x)

#set up a subplot grid that has height 2 and width 1,
#and set the first subplot as active
plt.subplot(2,1,1)

#make first plot
plt.plot(x, y_sin)
plt.title('Sine')

#set the second subplot as active and make the second plot
plt.subplot(2,1,2)
plt.plot(x, y_cos, 'g--')
plt.title('Cosine')

#show the figure
plt.show()
```

```python
#bar plot

x=[2,8,10]
y=[11,16,9]

x2=[3,9,11]
y2=[6,15,7]
plt.bar(x,y)
plt.bar(x2 ,y2, color='y')
plt.title('Bar graph')
plt.ylabel('Y-axis')
plt.xlabel('X-axis')
```

```python
#histograms

a=np.array([22, 87, 5, 43, 56, 73, 55, 54, 11, 20, 51, 5, 79, 31, 27])
plt.hist(a)
plt.title('Histogram')
plt.show()
```

```python
data=[np.random.normal(0, std, 100) for std in range(1,4)]
#range(1, 4): Generates numbers 1, 2, and 3.
#np.random.normal(0, std, 100): Generates 100 random numbers using a normal (bell-curve) distribution.
#The first argument (0) is the mean (center of the curve).
#The second argument (std) is the standard deviation (spread of the data).
#The third argument (100) is the sample size.List Comprehension: The outer brackets [...] combine these into a list of three separate datasets. Dataset 1 has a narrow spread (std=1), Dataset 2 has a medium spread (std=2), and Dataset 3 has a wide spread (std=3).


#rectangular boxplot
plt.boxplot(data, vert=True, patch_artist=True)
plt.show()

#data: Matplotlib detects three separate arrays in your list and automatically draws three separate boxes.
#vert=True: Forces the boxes to stand vertically. The values are read on the Y-axis.
#patch_artist=True: This changes the internal rendering engine from drawing simple lines (Line2D) to filled shapes (Patch). This is what allows the boxes to be filled with a solid color instead of remaining hollow.
```

```python
#data to plot
labels='Python', 'C++', 'Ruby', 'Java'
sizes=[215, 130, 245, 210]
colors=['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode=(0.1, 0, 0, 0) #explode 1st slice

#plot
plt.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', shadow=True)

plt.axis('equal')
plt.show()
```
