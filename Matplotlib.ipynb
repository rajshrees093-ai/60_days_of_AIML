It is a plotting lbrary for the puython programming language and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, Qt or GTK+

Some of the major Pros of Matplotlib are:


*   Generally easy to get started for simple plots
*   Support for custom labels and texts
*   Great ctrl of every element in a figure
*   High quality output in many formats
*   Very customizable in general








```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
import pandas as pd
```

```python
#simple examples

x=np.arange(0,10)
y=np.arange(11,21)
```

```python
a=np.arange(40,50)
b=np.arange(50,60)
```

```python
#plotting using matplotlib
#plt scatter

plt.scatter(x,y,c='g')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Graph in 20')

```

```python
y=x*x
```

```python
#plt plot
plt.plot(x,y, 'r*', linewidth=2, linestyle='dashed', markersize=12)
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('2-D diagram')
```

```python
#creating subplots
plt.subplot(2,2,1)
plt.plot(x,y,'r')
plt.subplot(2,2,2)
plt.plot(x,y,'g*--')
plt.subplot(2,2,3)
plt.plot(x,y, 'bo')





```

```python
x=np.arange(0,4*np.pi,0.1)
y=np.sin(x)
plt.title("Sine wave form")

#plot the points using matplotlib
plt.plot(x,y)
plt.show()
```

```python
#subplot()
#ompute the x and y cordinates for points on sine and cosine curves

x=np.arange(0, 5*np.pi, 0.1)
y_sin=np.sin(x)
y_cos=np.cos(x)

#set up a subplot grid that has height 2 and width 1,
#and set the first subplot as active
plt.subplot(2,1,1)

#make first plot
plt.plot(x, y_sin)
plt.title('Sine')

#set the second subplot as active and make the second plot
plt.subplot(2,1,2)
plt.plot(x, y_cos, 'g--')
plt.title('Cosine')

#show the figure
plt.show()
```

```python
#bar plot

x=[2,8,10]
y=[11,16,9]

x2=[3,9,11]
y2=[6,15,7]
plt.bar(x,y)
plt.bar(x2 ,y2, color='y')
plt.title('Bar graph')
plt.ylabel('Y-axis')
plt.xlabel('X-axis')
```
