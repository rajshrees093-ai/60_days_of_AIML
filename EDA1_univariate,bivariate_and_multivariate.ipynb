# UNIVARIATE, BIVARIATE AND MULTIVARIATE ANALYSIS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import kagglehub
path = kagglehub.dataset_download("arshid/iris-flower-dataset")
```

```python
import os
import pandas as pd

# Search for 'IRIS.csv' within the downloaded directory structure
iris_file_found = False
for root, dirs, files in os.walk(path):
    if 'IRIS.csv' in files: # Changed to 'IRIS.csv'
        file_path = os.path.join(root, 'IRIS.csv') # Changed to 'IRIS.csv'
        iris_file_found = True
        print(f"Found IRIS.csv at: {file_path}") # Added for debugging
        break

if not iris_file_found:
    raise FileNotFoundError("IRIS.csv not found within the downloaded dataset structure.")

df = pd.read_csv(file_path)
df.head()
```
